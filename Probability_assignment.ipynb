## Problem 1(a): Experimental Probability of Coin Tosses

### Objective
To simulate 10,000 coin tosses using Python and calculate the experimental probabilities of obtaining Heads and Tails.

### Problem Statement
Simulate 10,000 coin tosses and calculate the experimental probability of getting Heads and Tails.

### Concept
Experimental probability is the probability estimated from the outcomes observed during an experiment.

For this experiment:

Probability of Heads = Number of Heads / Total Tosses

Probability of Tails = Number of Tails / Total Tosses

### Approach

1. Import Python's `random` module.
2. Initialize counters for Heads and Tails to zero.
3. Simulate 10,000 coin tosses using a loop.
4. Randomly generate either Heads or Tails for each toss.
5. Update the appropriate counter based on the outcome.
6. Calculate the experimental probability of each outcome by dividing its count by the total number of tosses.
7. Display the counts and calculated probabilities.

**Note:** Since the tosses are random, the experimental probabilities may vary slightly between executions.

```python
import random

# Initialize variables
total_tosses = 10_000
heads = 0
tails = 0

# Simulate coin tosses
for _ in range(total_tosses):
    toss = random.choice(["Heads", "Tails"])

    if toss == "Heads":
        heads += 1
    else:
        tails += 1

# Calculate experimental probabilities
prob_heads = heads / total_tosses
prob_tails = tails / total_tosses

# Display results
print("Total tosses:", total_tosses)
print("Heads count:", heads)
print("Tails count:", tails)
print(f"Experimental probability of Heads: {prob_heads:.4f}")
print(f"Experimental probability of Tails: {prob_tails:.4f}")
```


### 📊 Result and Conclusion

The simulation estimates the **experimental probabilities of obtaining Heads and Tails** from **10,000 coin tosses**.

The probabilities are calculated by dividing the observed count of each outcome by the total number of tosses.

> **Theoretical Probability:** For a fair coin, the probability of obtaining Heads or Tails is **0.5 (50%)** for each outcome.

> **Experimental Probability:** The estimated probabilities are expected to be close to the theoretical probabilities. However, small differences may occur due to **random variation**.

#### ✅ Verification

- The total number of Heads and Tails should equal **10,000**.
- The sum of the experimental probabilities should equal **1 (100%)**.

#### 📝 Conclusion

The simulation demonstrates that **experimental probability** can be estimated by observing outcomes over repeated trials. As the number of tosses increases, the experimental probabilities generally tend to approach the theoretical probabilities.



## Problem 1(b): Probability of Getting a Sum of 7 with Two Dice

### **Problem Statement**
Roll two fair six-sided dice and estimate the probability of obtaining a sum of **7** using simulation.

### **Objective**
To simulate repeated rolls of two dice and calculate the **experimental probability** of getting a sum of 7.

### **Concept**
Each trial consists of rolling both dice once and adding their values. A trial is considered successful if the sum equals 7.

**Experimental Probability** = Successful Trials / Total Trials



### **Approach**

1. Import Python's `random` module.
2. Initialize a counter to track successful trials.
3. Perform **10,000 trials** using a loop.
4. Generate a random integer between 1 and 6 for each die.
5. Calculate the sum of the two dice.
6. If the sum equals 7, increment the success counter.
7. Calculate the experimental probability by dividing successful trials by total trials.
8. Display the number of successful trials and the estimated probability.


```python
import random

# Initialize variables
total_trials = 10_000
successes = 0

# Simulate rolling two dice
for _ in range(total_trials):
    die1 = random.randint(1, 6)
    die2 = random.randint(1, 6)

    if die1 + die2 == 7:
        successes += 1

# Calculate experimental probability
probability = successes / total_trials

# Display results
print("Total trials:", total_trials)
print("Successful trials (sum = 7):", successes)
print(f"Experimental probability: {probability:.4f}")
print(f"Experimental probability (%): {probability * 100:.2f}%")
```


### **📊 Result and Conclusion**

The simulation was performed over **10,000 trials**, with each trial consisting of one roll of two fair six-sided dice.

- **Total Trials:** 10,000
- **Successful Trials:** Number of trials in which the sum was 7.
- **Experimental Probability:** Successful Trials / Total Trials

> **Theoretical Probability:** The probability of obtaining a sum of 7 with two fair dice is **6/36 = 1/6 ≈ 0.1667 (16.67%)**.

The experimental probability may differ slightly from the theoretical probability due to **random variation**.

#### **📝 Conclusion**

The simulation demonstrates how Python can be used to estimate the probability of an event through repeated random experiments. With a sufficiently large number of trials, the experimental probability is generally expected to approach the theoretical probability.



## Problem 2: Probability of Getting at Least One 6

### **Problem Statement**
Write a Python function to estimate the probability of obtaining **at least one 6 in 10 rolls of a fair die** using simulation.

### **Objective**
To estimate the probability by repeating the experiment multiple times and calculating the proportion of trials in which at least one 6 occurs.

### **Concept**
A trial consists of rolling a fair die 10 times.

- **Success:** At least one roll produces a 6.
- **Failure:** None of the 10 rolls produces a 6.

**Experimental Probability** = Successful Trials / Total Trials



### **Approach**

1. Import Python's `random` module.
2. Define a function named `estimate_probability()`.
3. Accept the number of rolls per trial and the total number of trials as parameters.
4. Initialize a counter for successful trials.
5. For each trial, reset a Boolean flag to `False`.
6. Roll the die the specified number of times. If a 6 appears, set the flag to `True`.
7. After the rolls are complete, increment the success counter if the flag is `True`.
8. Calculate and return the experimental probability.
9. Call the function and display the estimated probability.


```python
import random

def estimate_probability(rolls_per_trial, total_trials):
    successes = 0

    for _ in range(total_trials):
        found_six = False

        for _ in range(rolls_per_trial):
            roll = random.randint(1, 6)

            if roll == 6:
                found_six = True

        if found_six:
            successes += 1

    probability = successes / total_trials
    return probability


# Estimate the probability
estimated_probability = estimate_probability(
    rolls_per_trial=10,
    total_trials=10_000
)

print(f"Estimated probability: {estimated_probability:.4f}")
print(f"Estimated probability (%): {estimated_probability * 100:.2f}%")
```


### **📊 Result and Conclusion**

The simulation estimates the probability of obtaining **at least one 6 in 10 rolls of a fair die** by repeating the experiment **10,000 times**.

The estimated probability is calculated as the proportion of successful trials among all simulated trials.

> **Theoretical Probability:** The probability of obtaining at least one 6 in 10 independent rolls is approximately **0.8385 (83.85%)**.

The experimental estimate may differ slightly from the theoretical value due to **random variation**.

#### **📝 Conclusion**

The simulation demonstrates how repeated random experiments can be used to estimate the probability of an event. By using a reusable Python function, the experiment can be repeated with different numbers of rolls and trials.


## Problem 3: Conditional Probability and Bayes' Theorem

### **Problem Statement**

A bag contains **5 red, 7 green, and 8 blue balls**. A ball is drawn randomly, its colour is recorded, and the ball is replaced. This process is repeated **1,000 times**.

1. Estimate the probability of drawing a red ball given that the previous draw was blue.
2. Verify **Bayes' theorem** using simulation.

### **Objective**

To estimate conditional probabilities through simulation and verify Bayes' theorem using the experimental results.

### **Concept**

**Conditional Probability** is the probability of an event occurring given that another event has already occurred.

For this experiment, we estimate the probability of drawing a Red ball on the current draw, given that the previous draw was Blue.

The formula is:

**P(Red | Previous Blue)** = Number of Blue-to-Red pairs / Total pairs with Previous Blue

Since the ball is replaced after every draw, the bag's composition remains unchanged. Therefore, the theoretical probability of drawing Red is 5/20 = **0.25 (25%)**.

### **Approach — Part (a)**

1. Import the `random` module.
2. Create a list containing 5 Red, 7 Green, and 8 Blue balls.
3. Draw the first ball and store its colour.
4. Simulate the remaining 999 draws.
5. Count the pairs where the previous ball was Blue.
6. Count how many of these pairs have Red as the current ball.
7. Calculate the estimated conditional probability by dividing the Blue-to-Red pairs by the pairs with Previous Blue.

```python
import random

# Create the bag containing 20 balls
bag = ["Red"] * 5 + ["Green"] * 7 + ["Blue"] * 8

total_draws = 1000

# Draw the first ball
previous_ball = random.choice(bag)

# Initialize counters
previous_blue = 0
blue_then_red = 0

# Simulate the remaining 999 draws
for _ in range(total_draws - 1):
    current_ball = random.choice(bag)

    if previous_ball == "Blue":
        previous_blue += 1

        if current_ball == "Red":
            blue_then_red += 1

    # Update the previous ball
    previous_ball = current_ball

# Calculate conditional probability
if previous_blue > 0:
    conditional_probability = blue_then_red / previous_blue

    print("Total draws:", total_draws)
    print("Pairs with Previous Blue:", previous_blue)
    print("Pairs with Blue followed by Red:", blue_then_red)
    print(f"Estimated P(Red | Previous Blue): {conditional_probability:.4f}")
else:
    print("No pairs had Blue as the previous ball.")
    
```

### **Result and Conclusion — Part (a)**

The simulation estimates the conditional probability of drawing a Red ball given that the previous draw was Blue.

**Estimated P(Red | Previous Blue)** = Blue-to-Red pairs / Pairs with Previous Blue

**Theoretical Probability:**

Since the draws are made with replacement, each draw has the same probability of producing a Red ball.

P(Red | Previous Blue) = 5/20 = **0.25 (25%)**

The experimental result may differ slightly from 25% because of random variation.

**Conclusion:** The simulation demonstrates how conditional probability can be estimated using consecutive random observations. Since the draws are independent, the previous draw being Blue does not change the probability of drawing Red on the current draw.

### **Part (b): Verification of Bayes' Theorem Using Simulation**

**Bayes' Theorem** describes the relationship between conditional probabilities.

The formula is:

**P(A | B) = [P(B | A) × P(A)] / P(B)**

Where:

- **A:** The current draw is Red.
- **B:** The previous draw was Blue.
- **P(A | B):** Probability that the current draw is Red, given that the previous draw was Blue.
- **P(B | A):** Probability that the previous draw was Blue, given that the current draw is Red.
- **P(A):** Probability that the current draw is Red.
- **P(B):** Probability that the previous draw was Blue.

**Objective:** To estimate these probabilities through simulation and verify that both sides of Bayes' theorem give approximately the same result.

### **Approach — Part (b)**

1. Create the bag containing 5 Red, 7 Green, and 8 Blue balls.
2. Simulate 10,000 consecutive pairs of draws, using replacement after each draw.
3. Count the pairs where the current draw is Red, the previous draw is Blue, and both events occur together.
4. Estimate the required probabilities from these counts.
5. Calculate the left-hand side, P(A | B).
6. Calculate the right-hand side, [P(B | A) × P(A)] / P(B).
7. Compare both values to verify Bayes' theorem.

```python
import random

# Create the bag
bag = ["Red"] * 5 + ["Green"] * 7 + ["Blue"] * 8

total_pairs = 10_000

# Initialize counters
current_red = 0
previous_blue = 0
blue_then_red = 0

# Simulate consecutive pairs of draws
for _ in range(total_pairs):
    previous_ball = random.choice(bag)
    current_ball = random.choice(bag)

    if current_ball == "Red":
        current_red += 1

    if previous_ball == "Blue":
        previous_blue += 1

    if previous_ball == "Blue" and current_ball == "Red":
        blue_then_red += 1

# Estimate probabilities
P_A = current_red / total_pairs
P_B = previous_blue / total_pairs

if previous_blue > 0 and current_red > 0 and P_B > 0:
    P_A_given_B = blue_then_red / previous_blue
    P_B_given_A = blue_then_red / current_red

    # Bayes' theorem
    bayes_result = (P_B_given_A * P_A) / P_B

    print(f"P(A): {P_A:.4f}")
    print(f"P(B): {P_B:.4f}")
    print(f"P(A | B): {P_A_given_B:.4f}")
    print(f"P(B | A): {P_B_given_A:.4f}")
    print(f"Bayes' theorem result: {bayes_result:.4f}")
    print(f"Difference: {abs(P_A_given_B - bayes_result):.6f}")
else:
    print("Insufficient observations to calculate the probabilities.")
```

### **Result and Conclusion — Part (b)**

The simulation estimates the probabilities required to verify Bayes' theorem.

**Theoretical values:**

- P(A) = P(Red) = 5/20 = 0.25
- P(B) = P(Previous Blue) = 8/20 = 0.40
- P(A | B) = 0.25
- P(B | A) = 0.40

Applying Bayes' theorem:

P(A | B) = [P(B | A) × P(A)] / P(B)

P(A | B) = (0.40 × 0.25) / 0.40 = **0.25**

**Conclusion:** The simulated values of both sides of Bayes' theorem should be approximately equal. Small differences may occur due to random variation. This verifies Bayes' theorem experimentally using simulated draws with replacement.

## Problem 4: Mean, Variance, and Standard Deviation

### **Problem Statement**

Generate a sample of **1,000 observations** from a discrete random variable X with the following probability distribution:

- P(X = 1) = 0.25
- P(X = 2) = 0.35
- P(X = 3) = 0.40

Calculate the **empirical mean, variance, and standard deviation** of the generated sample.

### **Objective**

To simulate a discrete random variable and estimate its statistical properties using Python.

### **Concept**

A **discrete random variable** takes specific, countable values, each with an associated probability.

- **Mean:** Represents the average value of the random variable.
- **Variance:** Measures how far the observations are spread around the mean.
- **Standard Deviation:** Represents the typical spread of observations around the mean.

The theoretical mean is calculated as:

Mean = Σ [x × P(X = x)]

The theoretical variance is calculated as:

Variance = Σ [P(X = x) × (x − Mean)²]

Standard Deviation = √Variance

The empirical values are calculated from the simulated sample and may differ slightly from the theoretical values due to random variation.

### **Approach**

1. Import NumPy.
2. Define the possible values of X: 1, 2, and 3.
3. Define their corresponding probabilities: 0.25, 0.35, and 0.40.
4. Generate 1,000 random observations using the specified probability distribution.
5. Calculate the sample mean using `np.mean()`.
6. Calculate the empirical variance using `np.var()`.
7. Calculate the standard deviation using `np.std()`.

```python
import numpy as np

# Define the values and their probabilities
values = [1, 2, 3]
probabilities = [0.25, 0.35, 0.40]

# Generate 1,000 observations
sample = np.random.choice(
    values,
    size=1000,
    p=probabilities
)

# Calculate empirical statistics
empirical_mean = np.mean(sample)
empirical_variance = np.var(sample)
empirical_std = np.std(sample)

# Display results
print("Sample size:", len(sample))
print(f"Empirical Mean: {empirical_mean:.4f}")
print(f"Empirical Variance: {empirical_variance:.4f}")
print(f"Empirical Standard Deviation: {empirical_std:.4f}")
```

### **Theoretical Calculations**

The theoretical mean is:

Mean = (1 × 0.25) + (2 × 0.35) + (3 × 0.40)

Mean = 0.25 + 0.70 + 1.20 = **2.15**

The theoretical variance is:

Variance = [0.25 × (1 − 2.15)²] + [0.35 × (2 − 2.15)²] + [0.40 × (3 − 2.15)²]

Variance = 0.330625 + 0.007875 + 0.289 = **0.6275**

Standard Deviation = √0.6275 = **0.7922 approximately**

### **Result and Conclusion**

The simulation generates 1,000 observations from the given discrete probability distribution and calculates their empirical mean, variance, and standard deviation.

The theoretical values are:

- **Mean:** 2.15
- **Variance:** 0.6275
- **Standard Deviation:** Approximately 0.7922

The empirical results should be reasonably close to these theoretical values. Small differences are expected because the sample is generated randomly.

**Conclusion:** This experiment demonstrates how NumPy can simulate a discrete random variable and estimate its statistical properties.

## Problem 5: Exponential Distribution

### **Problem Statement**

Simulate **2,000 samples** from an exponential distribution with a mean of 5. Plot a histogram of the simulated data and overlay the theoretical Probability Density Function (PDF).

### **Objective**

To understand the exponential distribution through simulation and visualize how the simulated data compares with its theoretical probability density function.

### **Concept**

The **Exponential Distribution** is a continuous probability distribution commonly used to model waiting times between independent events occurring at a constant average rate.

Examples include the waiting time between customer arrivals or between certain system events.

The exponential distribution is defined by its **scale parameter**, which is equal to its mean.

For this problem:

- Mean = 5
- Scale parameter = 5
- Rate parameter = 1/5 = 0.2

The theoretical Probability Density Function (PDF) is:

**f(x) = (1/5) × exp(−x/5), for x ≥ 0**

A histogram displays the distribution of simulated observations. Overlaying the theoretical PDF allows us to compare the simulated data with the expected distribution.

### **Approach**

1. Import NumPy, Matplotlib, and the required plotting functions.
2. Generate 2,000 random observations from an exponential distribution with a scale of 5.
3. Plot a density-normalized histogram of the observations.
4. Calculate the theoretical PDF for a range of x-values.
5. Overlay the theoretical PDF on the histogram.
6. Add a title, axis labels, legend, and grid to make the visualization clear.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
mean = 5
sample_size = 2000

# Generate samples from the exponential distribution
samples = np.random.exponential(
    scale=mean,
    size=sample_size
)

# Create x-values for the theoretical PDF
x = np.linspace(0, samples.max(), 500)

# Calculate the theoretical PDF
pdf = (1 / mean) * np.exp(-x / mean)

# Plot the histogram
plt.figure(figsize=(10, 6))
plt.hist(
    samples,
    bins=30,
    density=True,
    alpha=0.6,
    edgecolor="black",
    label="Simulated Data"
)

# Overlay the theoretical PDF
plt.plot(
    x,
    pdf,
    linewidth=2,
    label="Theoretical PDF"
)

# Add labels and formatting
plt.title("Exponential Distribution (Mean = 5)")
plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### **Result and Conclusion**

The simulation generates 2,000 observations from an exponential distribution with a mean of 5. The histogram represents the simulated data, while the curve represents the theoretical probability density function.

**Observations:**

- Most observations are concentrated near zero.
- The probability density decreases as the value increases.
- Larger waiting times are possible but become progressively less likely.
- The histogram should approximately follow the theoretical PDF, although small differences are expected because of random variation.

**Conclusion:** This experiment demonstrates how to simulate a continuous probability distribution and compare the empirical distribution with its theoretical PDF using Python visualization tools.

## Problem 6: Central Limit Theorem (CLT)

### **Problem Statement**

Generate **10,000 random numbers** from a uniform distribution. Draw **1,000 samples**, each containing 30 observations, and calculate the mean of each sample. Visualize the distribution of these sample means.

### **Objective**

To demonstrate the Central Limit Theorem by observing how the distribution of sample means approaches a normal distribution as the sample size increases.

### **Concept**

The **Central Limit Theorem (CLT)** states that, under suitable conditions, the distribution of sample means approaches a normal distribution as the sample size becomes sufficiently large, even when the original population is not normally distributed.

In this experiment:

- Population distribution: Uniform distribution
- Population size: 10,000 random numbers
- Number of samples: 1,000
- Sample size: 30 observations
- Statistic calculated: Mean of each sample

For a uniform distribution between 0 and 1:

- Population mean = 0.5
- Population variance = 1/12
- Standard error of the sample mean = √(1/12) / √30

The histogram of the 1,000 sample means should be approximately bell-shaped and centered near 0.5.

### **Approach**

1. Import NumPy and Matplotlib.
2. Generate 10,000 random numbers from a uniform distribution between 0 and 1.
3. Draw 1,000 samples, each containing 30 observations, from the generated population.
4. Calculate the mean of each sample.
5. Plot a histogram of the sample means.
6. Add a vertical line representing the theoretical population mean of 0.5.
7. Label the plot and interpret the resulting distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate the population
population = np.random.uniform(
    low=0,
    high=1,
    size=10_000
)

# Define simulation parameters
num_samples = 1000
sample_size = 30

# Generate samples and calculate their means
sample_means = []

for _ in range(num_samples):
    sample = np.random.choice(
        population,
        size=sample_size,
        replace=True
    )
    sample_means.append(np.mean(sample))

sample_means = np.array(sample_means)

# Calculate the overall mean of the sample means
mean_of_sample_means = np.mean(sample_means)

# Plot the distribution of sample means
plt.figure(figsize=(10, 6))

plt.hist(
    sample_means,
    bins=30,
    density=True,
    edgecolor="black",
    alpha=0.7,
    label="Distribution of Sample Means"
)

plt.axvline(
    x=0.5,
    linestyle="--",
    linewidth=2,
    label="Theoretical Population Mean = 0.5"
)

plt.title("Central Limit Theorem (Sample Size = 30)")
plt.xlabel("Sample Mean")
plt.ylabel("Density")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print(f"Number of samples: {num_samples}")
print(f"Sample size: {sample_size}")
print(f"Mean of sample means: {mean_of_sample_means:.4f}")
print(f"Theoretical population mean: {0.5:.4f}")
```

### **Result and Conclusion**

The simulation generates 1,000 samples, each containing 30 observations, from a uniform population. The mean of each sample is calculated and visualized using a histogram.

**Observations:**

- The sample means are concentrated around 0.5.
- The distribution of sample means is approximately bell-shaped.
- The variability of sample means is smaller than the variability of individual population observations.
- The simulated mean of the sample means should be close to the theoretical population mean of 0.5.

**Conclusion:** The experiment illustrates the Central Limit Theorem. Even though the original population follows a uniform distribution, the distribution of sample means becomes approximately normal for a sufficiently large sample size.
